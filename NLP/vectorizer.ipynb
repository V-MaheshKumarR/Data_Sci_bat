```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
d0 = 'i bought iphone today not good'
d1 = 'iphone experience is very bad'
d2 = 'android is better than iphone'
d3 = 'iphone is good'

Corpus = [d0, d1, d2,d3]
```

```python
tfidf = TfidfVectorizer()
result = tfidf.fit_transform(Corpus)

```

```python
print('\nidf values:')
for ele1, ele2 in zip(tfidf.get_feature_names_out(), tfidf.idf_):
    print(ele1, ':', ele2)
```
