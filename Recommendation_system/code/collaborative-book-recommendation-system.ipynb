#### Cluster users by their rating behaviour.
##### user_matrix: (n_users × n_books) — transpose of pivot.
    Returns:
      labels       np.ndarray  cluster assignment per user (index = row of user_matrix)
      user_cluster dict        {user_id → cluster_id}
      km           KMeans      fitted model
      user_matrix  DataFrame   (users × books)

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity


user_matrix = pt.T                                  # users × books
scaler      = StandardScaler()
X           = scaler.fit_transform(user_matrix.fillna(0))
km          = KMeans(n_clusters=5, init="k-means++",n_init=15, random_state=42, max_iter=300)
labels = km.fit_predict(X)
user_cluster = dict(zip(user_matrix.index, labels))
```

```python
km
```

```python
user_cluster_ser = pd.Series(user_cluster)
user_cluster_ser
```

#### Assign readable labels to clusters based on which books (by index position) each cluster rates highest on average.

```python
cluster_labels = user_cluster_ser.values

labels = {}
um = user_matrix.copy()
um["cluster"] = cluster_labels

for c in range(cluster_labels.max() + 1):
        members  = um[um["cluster"] == c].drop(columns="cluster")
        avg_rats = members.mean()
        top_books = avg_rats.sort_values(ascending=False).head(2).index.tolist()
        # Shorten to first 3 words of each title
        short = [" ".join(t.split()[:3]) for t in top_books]
        labels[c] = " & ".join(short) if short else f"Cluster {c}"
```

# SECTION 5 — USER-BASED RECOMMENDATION (unread books only)
# ══════════════════════════════════════════════════════════════════


```python
def recommend_for_user(user_id: int,
                        user_matrix: pd.DataFrame,
                        user_cluster: dict,
                        books: pd.DataFrame,
                        ratings: pd.DataFrame,
                        top_n: int = 8) -> dict:
    """
    Recommend unread books for user_id.

    Algorithm
    ---------
    1. Identify the user's K-Means cluster.
    2. Gather cluster peers (other users in same cluster).
    3. Compute cosine similarity between target and each peer.
    4. For every unread book, compute:
         CF score  = Σ(sim_i × rating_i) / Σ|sim_i|   (peer-weighted avg)
         pop score = mean_global_rating / 10
         final     = 0.70 × CF/10 + 0.30 × pop_score
    5. Return top_n books ranked by final score,
       filtering out any book the user has already read.
    """
    if user_id not in user_matrix.index:
        return {"error": f"User {user_id} not in filtered dataset. "
                         f"They may not have ≥10 ratings. "
                         f"Try: {list(user_matrix.index[:5])}"}

    uid_cluster = user_cluster[user_id]
    peer_ids    = [u for u, c in user_cluster.items()
                   if c == uid_cluster and u != user_id]
    if not peer_ids:
        peer_ids = [u for u in user_matrix.index if u != user_id]

    # Identify unread books
    user_row    = user_matrix.loc[user_id]
    read_titles = set(user_row[user_row > 0].index)
    unread      = [b for b in user_matrix.columns if b not in read_titles]

    if not unread:
        return {"error": "User has rated every book in the catalogue!"}

    # Cosine similarity with peers
    target_vec   = user_row.values.reshape(1, -1)
    peer_matrix  = user_matrix.loc[peer_ids]
    sims         = cosine_similarity(target_vec, peer_matrix)[0]

    # Global average rating per book (from raw ratings + books)
    isbn_to_title = books.set_index("ISBN")["title"].to_dict()
    title_to_isbn = {v: k for k, v in isbn_to_title.items()}

    scores = []
    for title in unread:
        peer_rats = peer_matrix[title].values
        rated_m   = peer_rats > 0
        if rated_m.sum() == 0:
            cf = 0.0
        else:
            w  = sims[rated_m]; r = peer_rats[rated_m]
            cf = float(np.dot(w, r) / (np.abs(w).sum() + 1e-9))

        isbn = title_to_isbn.get(title, "")
        if isbn:
            raw_avg = ratings[ratings["ISBN"] == isbn]["rating"].mean()
            pop = float(raw_avg) / 10.0 if not np.isnan(raw_avg) else 0.5
        else:
            pop = 0.5

        final = 0.70 * (cf / 10.0) + 0.30 * pop

        meta = books[books["title"] == title]
        scores.append({
            "title":      title,
            "author":     meta["author"].values[0]    if len(meta) else "Unknown",
            "year":       int(meta["year"].values[0]) if len(meta) and pd.notna(meta["year"].values[0]) else "N/A",
            "publisher":  meta["publisher"].values[0] if len(meta) and "publisher" in meta.columns else "",
            "ISBN":       isbn,
            "cf_score":   round(cf, 3),
            "pop_score":  round(pop * 10, 2),
            "match_pct":  round(final * 100, 1),
        })

    scores.sort(key=lambda x: x["match_pct"], reverse=True)

    # User top genres (from raw ratings)
    user_ratings_df = ratings[ratings["user_id"] == user_id]
    user_books_df   = user_ratings_df.merge(books, on="ISBN")
    top_genres      = {}
    if "genre" in user_books_df.columns:
        top_genres = (
            user_books_df.groupby("genre")["rating"].mean()
            .sort_values(ascending=False).head(3).round(1).to_dict()
        )

    return {
        "user_id":         user_id,
        "cluster":         int(uid_cluster),
        "cluster_size":    sum(1 for c in user_cluster.values() if c == uid_cluster),
        "books_read":      len(read_titles),
        "top_genres":      top_genres,
        "sample_read":     list(read_titles)[:5],
        "recommendations": scores[:top_n],
    }
```

```python
import os
import pickle

def save_model(path: str, **kwargs):
    # path = r'D:\Data Science_bat\Recommendation_system\models'
    os.makedirs(os.path.dirname(path) or ".", exist_ok=True)
    with open(path, "wb") as f:
        pickle.dump(kwargs, f)

def load_model(path: str) -> dict:
    with open(path, "rb") as f:
        return pickle.load(f)
```

```python
import textwrap
CLUSTER_ICONS = ["🦁","🦋","🦅","🐬","🦊"]

def print_banner():
    print("\n" + "═"*65)
    print("  📚  BookMind — Intelligent Book Recommendation System")
    print("      K-Means Clustering  +  Collaborative Filtering")
    print("      Dataset: Kaggle Book-Crossing (Books / Ratings / Users)")
    print("═"*65)

def _bar(pct, width=20):
    filled = int(pct / 100 * width)
    return "█"*filled + "░"*(width-filled)

def print_cluster_report(cluster_labels, cluster_label_map, user_matrix):
    print("\n🧩  CLUSTER ANALYSIS")
    print("─"*65)
    n = cluster_labels.max() + 1
    for c in range(n):
        size  = int((cluster_labels == c).sum())
        label = cluster_label_map[c]
        icon  = CLUSTER_ICONS[c] if c < len(CLUSTER_ICONS) else "📚"
        bar   = _bar(size / len(cluster_labels) * 100, width=25)
        print(f"\n  {icon}  Cluster {c}  │  {label}")
        print(f"     {bar}  {size} users ({size/len(cluster_labels)*100:.1f}%)")

def print_user_recs(result: dict, cluster_label_map: dict):
    if "error" in result:
        print(f"\n  ❌  {result['error']}")
        return

    uid     = result["user_id"]
    cid     = result["cluster"]
    label   = cluster_label_map.get(cid, f"Cluster {cid}")
    icon    = CLUSTER_ICONS[cid] if cid < len(CLUSTER_ICONS) else "📚"

    print(f"\n{'═'*65}")
    print(f"  👤  User #{uid}   {icon} {label}")
    print(f"  📖  Books read: {result['books_read']:,}  │  Cluster peers: {result['cluster_size']:,}")

    if result["top_genres"]:
        print(f"\n  ❤   Favourite genres:")
        for g, avg in result["top_genres"].items():
            print(f"       {'• ' + g:<28} {avg:.1f} / 10 ★")

    if result["sample_read"]:
        print(f"\n  ✓   Recently read:")
        for t in result["sample_read"]:
            print(f"       {textwrap.shorten(t, 55)}")

    print(f"\n{'─'*65}")
    print(f"  ✨  TOP RECOMMENDATIONS  (unread books only)")
    print(f"{'─'*65}")

    for i, rec in enumerate(result["recommendations"], 1):
        bar = _bar(rec["match_pct"])
        print(f"\n  #{i:<2}  {textwrap.shorten(rec['title'], 50)}")
        print(f"       ✍  {rec['author']}  ({rec['year']})")
        if rec.get("publisher"):
            print(f"       🏢  {rec['publisher']}")
        if rec.get("ISBN"):
            print(f"       📎  ISBN: {rec['ISBN']}")
        print(f"       {bar}  {rec['match_pct']:.1f}% match")

    print(f"\n{'═'*65}\n")

def print_similar_books(title: str, recs: list):
    print(f"\n{'═'*65}")
    print(f"  🔍  Books similar to: {title}")
    print(f"{'─'*65}")
    if not recs:
        print(f"  ❌  '{title}' not found in filtered catalogue.")
        return
    for i, rec in enumerate(recs, 1):
        bar = _bar(rec["match_pct"])
        print(f"\n  #{i:<2}  {textwrap.shorten(rec['title'], 50)}")
        print(f"       ✍  {rec['author']}  ({rec['year']})")
        print(f"       {bar}  {rec['match_pct']:.1f}% similarity")
    print(f"\n{'═'*65}\n")
```

```python
def similar_books(book_title: str, pivot: pd.DataFrame,
                  similarity: np.ndarray, books: pd.DataFrame,
                  top_n: int = 6) -> list[dict]:
    """
    Item-based recommendation — identical logic to notebook's recommend().
    Returns top_n most similar books to book_title.
    """
    if book_title not in pivot.index:
        return []

    idx        = list(pivot.index).index(book_title)
    sim_scores = sorted(enumerate(similarity[idx]), key=lambda x: x[1], reverse=True)[1:top_n+1]

    results = []
    for i, score in sim_scores:
        title = pivot.index[i]
        meta  = books[books["title"] == title]
        results.append({
            "title":     title,
            "author":    meta["author"].values[0]    if len(meta) else "Unknown",
            "year":      int(meta["year"].values[0]) if len(meta) and pd.notna(meta["year"].values[0]) else "N/A",
            "publisher": meta["publisher"].values[0] if len(meta) and "publisher" in meta.columns else "",
            "score":     round(float(score), 4),
            "match_pct": round(float(score) * 100, 1),
        })
    return results
```
