## 1. FAISS - Storing Embeddings in a Vector Index

FAISS (Facebook AI Similarity Search) is an in-memory library for fast similarity search.

**Install:** `pip install faiss-cpu numpy`

```python
pip install faiss-cpu
```

```python
# pip install faiss-cpu numpy
import faiss
import numpy as np
from openai import OpenAI

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key= os.environ["OPENROUTER_API_KEY_3"],
)

EMBED_MODEL = "nvidia/nemotron-3-embed-1b:free",

def get_embedding(text: str):
    resp = client.embeddings.create(
        model=EMBED_MODEL,
        input= text,
        encoding_format="float",
    )
    return resp.data[0].embedding

# Sample documents (your source text / metadata)
documents = [
    "Python is a popular programming language for data science.",
    "FAISS is a library for efficient similarity search.",
    "Chroma is an open-source embedding database.",
    "Large Language Models can generate human-like text.",
    "Vector databases store numerical representations of data.",
]

# Generate embeddings for all documents
embeddings = []

for document in documents:
    print(document)
    embed = get_embedding(document)
    embeddings.append(embed)

embeddings_np = np.array(embeddings, dtype="float32")

print("Embeddings shape:", embeddings_np.shape)
```

```python
from openai import OpenAI

client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key= os.environ["OPENROUTER_API_KEY_3"],
)

embedding = client.embeddings.create(
  model="nvidia/nemotron-3-embed-1b:free",
  input=documents,
  encoding_format="float"
)
print(embedding.data[0].embedding)
```

```python
embedding
```

```python
# Build a FAISS index (L2 distance)  (vector index) like creating table in database
dimension = embeddings_np.shape[1]
index = faiss.IndexFlatL2(dimension) #create a index based on the dimensions of embedings
index.add(embeddings_np)  # add the embeddings on buit index

print("Total vectors in index:", index.ntotal)
```

```python
# Query the FAISS index
query = "tell me about programming language used in data science"
query_vec = np.array([get_embedding(query)], dtype="float32")

query_vec
```

```python
k = 3  # top-k results
distances, indices = index.search(query_vec, k)

```

```python
print(f"Query: {query}\n")
for rank, (idx, dist) in enumerate(zip(indices[0], distances[0]), start=1):
    print(f"{rank}. (distance={dist:.4f}) {documents[idx]}")
```

```python
# Save and load FAISS index to disk
faiss.write_index(index, "faiss_index.bin")

loaded_index = faiss.read_index("faiss_index.bin")
print("Loaded vectors:", loaded_index.ntotal)
```

## 2. Chroma DB - Storing Embeddings with Metadata

Chroma is a persistent vector database with built-in metadata support.

**Install:** `pip install chromadb`

```python
pip install chromadb
```

```python
import chromadb

# Persistent client (stores data on disk permanently)
chroma_client = chromadb.PersistentClient(path="./chroma_store")

# Create or get a collection

collection = chroma_client.get_or_create_collection(name="my_documents_new_1",configuration={"hnsw": {"space": "cosine"}})

print("Collection ready:", collection.name, configuration={"hnsw": {"space": "cosine"}})
```

```python
# Add documents with embeddings + metadata
ids = [f"doc_{i}" for i in range(len(documents))]
metadatas = [{"source": "manual", "doc_id": i, "text": documents[i]} for i in range(len(documents))]

metadatas
```

```python
ids = [f"doc_{i}" for i in range(len(documents))]
metadatas = [{'source': 'python',
  'doc_id': 0,
  'text': 'Python is a popular programming language for data science.'},
 {'source': 'vector db',
  'doc_id': 1,
  'text': 'FAISS is a library for efficient similarity search.'},
 {'source': 'vector db',
  'doc_id': 2,
  'text': 'Chroma is an open-source embedding database.'},
 {'source': 'LLM',
  'doc_id': 3,
  'text': 'Large Language Models can generate human-like text.'},
 {'source': 'vector db',
  'doc_id': 4,
  'text': 'Vector databases store numerical representations of data.'}]
```

```python
metadatas
```

```python
collection.add(
    ids=ids,
    embeddings=embeddings,        # reuse embeddings from FAISS section
    documents=documents,          # original text
    metadatas=metadatas,
)

print("Total documents in collection:", collection.count())
```

```python
# Query Chroma using an embedding
query = "tell me about programming language used in data science"
query_embedding = get_embedding(query)

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3,
)

results
```

```python
print(f"Query: {query}\n")
for doc, meta, dist in zip(results["documents"][0], results["metadatas"][0], results["distances"][0]):
    print(f"- (dist={dist:.4f}) {doc}  | meta={meta}")
```

```python
# Filter using metadata
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=2,
    where={"source": "python"},
)
print(query)
results
```
