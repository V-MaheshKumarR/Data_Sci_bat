```python

from pymongo import MongoClient
from pymongo.server_api import ServerApi

uri = os.environ["MONGODB_URI"]

# Create a new client and connect to the server
client = MongoClient(uri, server_api=ServerApi('1'))

collections = client["rag_new_db"]["war_documents"]
```

```python
from openai import OpenAI

client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=os.environ["OPENROUTER_API_KEY_2"],
)

def get_embeddings(chunk_text):

    # Image input embeddings use multimodal content format
    embedding = client.embeddings.create(
    model="nvidia/llama-nemotron-embed-vl-1b-v2:free",
    dimensions= 1024,
    input=[
        {
        "content": [
            {"type": "text", "text": chunk_text}
        ]
        }
    ],
    encoding_format="float"
    )

    return embedding.data[0].embedding
```

```python
chunk1_embedings = get_embeddings(chunks[0])

document = {
    "text": chunks[0],
    "embeddings": chunk1_embedings
}

collections.insert_one(document)
```

```python
query_embedding = get_embeddings("what are the Military clauses")
# Convert our search query into a vector (same way we did for documents)

results = collections.aggregate([
      {
        "$vectorSearch": {
          "index": "rag_index",
          "queryVector": query_embedding,
          "path": "embeddings",
          "numCandidates": 1024,
          "limit": 5
          # Step 2: Find the 5 most similar chunks in MongoDB
        }
      },
      {
        "$project": {
          "_id": 0,      # Exclude the MongoDB ID from results
          "text": 1      # Include only the "text" field
          # Step 3: Return just the text, not the embedding vectors
        }
      }
  ])
```

```python

array_of_results = []
for doc in results:
    array_of_results.append(doc)

array_of_results
```

### Treaties That Govern War  

When states go to war (or threaten to), **international law** steps in to limit the means and methods of conflict, protect those who are not (or no longer) participating, and lay out rules for ending hostilities.  
The major treaties that form the modern legal framework for armed conflict are:

| Treaty / Convention | Year(s) Adopted | Main Scope | Key Provisions |
|---------------------|----------------|------------|----------------|
| **Hague Conventions** (First & Second) | 1899, 1907 | Laws of war on the battlefield; conduct of hostilities; treatment of prisoners; means of warfare | • Prohibits unnecessary suffering (e.g., certain weapons). <br>• Requires distinction between combatants and civilians. <br>• Sets rules for the declaration of war, neutral rights, occupation, and the treatment of prisoners of war (POWs). |
| **Geneva Conventions** (Four Conventions + Additional Protocols) | 1949 (Conventions) ; 1977, 2005 (Protocols) | Protection of persons who are **not** or **no longer** taking part in hostilities (wounded, shipwrecked, prisoners, civilians). | • **1st Convention** – Wounded & sick in land forces.<br>• **2nd Convention** – Wounded & sick in naval forces.<br>• **3rd Convention** – Treatment of POWs.<br>• **4th Convention** – Protection of civilians in occupied territories.<br>• **Additional Protocol I** – International armed conflicts (rules on targeting, protected objects, combatant status).<br>• **Additional Protocol II** – Non‑international (civil) armed conflicts.<br>• **Protocol III** – Adoption of the Red Crystal emblem. |
| **Convention on Certain Conventional Weapons (CCW)** | 1980 (with multiple Amendments) | Limits or bans specific weapons deemed to cause unnecessary injury or have indiscriminate effects. | • Cluster munitions, landmines, incendiary weapons, blinding laser weapons, etc. <br>• Each “Protocol” adds a new weapon category or clarifies use‑restrictions. |
| **Chemical Weapons Convention (CWC)** | 1993 | Prohibits development, production, stockpiling, and use of chemical weapons. | • Destruction of existing stockpiles, verification regime (OPCW). |
| **Biological Weapons Convention (BWC)** | 1972 | Outlaws development, production, and stockpiling of biological agents for weapon use. | • No verification body (still a weakness), but a strong normative prohibition. |
| **Convention on the Prohibition of Anti‑Personnel Mines (Ottawa Treaty)** | 1997 | Bans the use, production, transfer, and stockpiling of landmines. | • Requires destruction of existing stockpiles and clearance of mined areas. |
| **Treaty on the Prohibition of Nuclear Weapons (TPNW)** | 2017 | First legally binding instrument to comprehensively prohibit nuclear weapons. | • Prohibits development, testing, production, acquisition, possession, stockpiling, use, and threats of use. |
| **Additional treaties & protocols relevant to war** | | | • **Treaty on the Non‑Proliferation of Nuclear Weapons (NPT)** – limits nuclear arms spread.<br>• **Arms Trade Treaty (ATT)** – regulates international transfer of conventional weapons.<br>• **UN Charter (1945)** – sets the legal basis for the *use* of force (self‑defence, Security Council authorisations).<br>• **Rome Statute of the International Criminal Court (1998)** – defines war crimes, crimes against humanity, genocide, and aggression. |

---

## 1. Why These Treaties Matter  

1. **Humanitarian Protection** – They create *minimum standards* for how parties must treat the wounded, POWs, and civilians.  
2. **Limiting Means & Methods** – Certain weapons (chemical, biological, cluster munitions, etc.) are banned or heavily regulated because of their indiscriminate or excessively cruel effects.  
3. **Accountability** – Violations can constitute **war crimes** or **crimes against humanity**, bringing individuals (and sometimes states) before international courts (ICC, ad hoc tribunals, national courts).  
4. **Influence on Conduct of War** – Commanders must plan operations within the legal limits (e.g., proportionality, distinction, precaution). Failure to do so can lead to legal liability and loss of legitimacy.

---

## 2. How the Treaties Interact  

1. **Hierarchy & Complementarity**  
   - The **UN Charter** is the *foundational* treaty governing the *use* of force.  
   - The **Hague** and **Geneva** Conventions govern *how* war is fought once it has begun.  
   - The **CCW**, **CWC**, **BWC**, **Ottawa Treaty**, etc., are *specialised* arms control instruments that sit alongside the broader humanitarian law regime.

2. **Universal vs. Regional/Selective Applicability**  
   - The four Geneva Conventions have **near‑universal ratification** (over 190 states).  
   - Some weapons bans (e.g., landmines, cluster munitions) have **majority but not universal** participation; major powers such as the United States, Russia, and China have not joined the Ottawa Treaty.  
   - The **TPNW** also has limited adoption; nuclear‑armed states have not joined.

3. **Customary International Law**  
   - Many provisions of the Hague and Geneva Conventions have become **customary law**, meaning they bind even non‑party states (e.g., the principle of distinction).  

---

## 3. Core Principles of Modern International Humanitarian Law (IHL)  

| Principle | Description | Treaty Source |
|-----------|-------------|----------------|
| **Distinction** | Parties must always distinguish between combatants (or other lawful targets) and civilians/civilian objects. | Geneva Conventions; Additional Protocol I |
| **Proportionality** | Attacks may not cause civilian harm “excessive in relation to the concrete and direct military advantage anticipated.” | Additional Protocol I |
| **Necessity** | Use of force must be limited to what is necessary to achieve a legitimate military objective. | General principle, reflected in Hague & Geneva |
| **Precaution** | Parties must take all feasible steps to minimize civilian harm (e.g., warnings, choosing less harmful means). | Additional Protocol I |
| **Humane Treatment** | No person taking no part in hostilities may be subjected to violence, torture, humiliating treatment, etc. | Geneva Conventions (esp. 3rd & 4th) |
| **Prohibition of Unnecessary Suffering** | Weapons that cause super‑fluous injury or unnecessary suffering are prohibited. | Hague Conventions; CCW protocols |

---

## 4. Enforcement & Compliance  

| Mechanism | How It Works | Example |
|-----------|--------------|---------|
| **International Criminal Court (ICC)** | Prosecutes individuals for war crimes, crimes against humanity, genocide, and the crime of aggression (if the state is a party or referred by the UN). | *prosecutor v. Al-Bashir* (ICC investigation) |
| **Ad Hoc Tribunals** | Created by the UN Security Council for specific conflicts (e.g., ICTY, ICTR). | *ICTY* prosecuted violations of the Geneva Conventions in the Balkans. |
| **Universal Jurisdiction** | Some states allow their courts to try war‑crime suspects regardless of where the crime occurred. | Spain’s “universal jurisdiction” cases. |
| **Compliance Monitoring Bodies** | OPCW (for CWC), IAEA (for nuclear safeguards), UN roving arms inspectors, NGOs (e.g., ICRC, Human Rights Watch) document violations. | OPCW’s inspections of Syrian chemical‑weapon sites. |
| **State‑to‑State Diplomatic Pressure & Sanctions** | UN Security Council resolutions can impose arms embargoes, sanctions, or authorise collective action. | UNSC Res. 1973 (Libya, 2011). |

## 5. Frequently Asked Questions  

| Question | Short Answer |
|----------|--------------|
| **Do the treaties apply to non‑state actors (e.g., rebel groups)?** | Yes. While treaties bind states, many of the core rules (distinction, prohibition of torture, protection of civilians) are **customary international law** and therefore bind non‑state parties as well. |
| **What happens if a state has not signed a particular treaty?** | It is still bound by customary IHL (e.g., the principle of distinction). However, it is not bound by the treaty‑specific obligations (e.g., the land‑mine ban if it never ratified the Ottawa Treaty). |
| **Can a country “opt out” of the Geneva Conventions during a conflict?** | No. The Conventions are *self‑executing* upon ratification and are considered *peremptory norms* (jus cogens). A state cannot unilaterally withdraw from core humanitarian obligations during hostilities. |
| **Are economic sanctions considered a “weapon” under these treaties?** | Sanctions are not covered by the arms‑control treaties, but a **comprehensive blockade** that deprives a civilian population of essential goods can, under IHL, be considered unlawful if it causes excessive suffering. |
| **Does the use of drones violate any treaty?** | Not per se. Drones are not banned, but their use must respect the same IHL principles (distinction, proportionality, precaution). Targeted killings of civilians or excessive civilian harm would constitute a violation. |

### Bottom Line  

- **War is never “lawless.”** A dense web of treaties—anchored by the UN Charter, the Hague Conventions, and the Geneva Conventions—sets out the *right* way to conduct hostilities and the *minimum* protections owed to those caught in the cross‑fire.  
- Violations can trigger **criminal liability**, **international sanctions**, and **political consequences**.  
- While not every country has signed every weapons‑control treaty, the core humanitarian principles have become **customary law**, binding on **all** parties—state and non‑state alike.
