```python
import os
from dotenv import find_dotenv, load_dotenv

# Install with: python -m pip install -r requirements-env.txt (from the repository root).
env_example = find_dotenv(".env.example", usecwd=True)
env_path = env_example.removesuffix(".example")
if not env_path or not os.path.isfile(env_path):
    raise FileNotFoundError("Create a local .env using the root .env.example first.")
load_dotenv(env_path)
```

```python
API_KEY = os.environ["OPENROUTER_API_KEY_3"]
```

```python
import requests
import json

# First API call with reasoning
response = requests.post(
  url="https://openrouter.ai/api/v1/chat/completions",
  headers={
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json",
  },
  data=json.dumps({
    "model": "apodex/apodex-1.1-mini:free",
    "messages": [
        {
          "role": "user",
          "content": "How many r's are in the word 'strawberry'?"
        }
      ],
    "reasoning": {"enabled": True}
  })
)

# Extract the assistant message with reasoning_details
response = response.json()
response = response['choices'][0]['message']
```

```python
from openai import OpenAI

client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key= os.environ["OPENROUTER_API_KEY_3"],
)

response = client.chat.completions.create(
  model="apodex/apodex-1.1-mini:free",
  messages=[
          {
            "role": "user",
            "content": "which countries partipated in the war ?"
          }
        ],
  extra_body={"reasoning": {"enabled": True}}
)

# Extract the assistant message with reasoning_details
response = response.choices[0].message
print(response)
```

```python
print(response.content)
```

```python
# Preserve the assistant message with reasoning_details
messages = [
  {"role": "user", "content": "How many r's are in the word 'strawberry'?"},
  {
    "role": "assistant",
    "content": response.content,
    "reasoning_details": response.reasoning_details  # Pass back unmodified
  },
  {"role": "user", "content": "Are you sure? Think carefully."}
]

# Second API call - model continues reasoning from where it left off
response2 = client.chat.completions.create(
  model="apodex/apodex-1.1-mini:free",
  messages=messages,
  extra_body={"reasoning": {"enabled": True}}
)
```
