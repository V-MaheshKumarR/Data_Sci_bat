```python
import os
from dotenv import find_dotenv, load_dotenv

# Install with: python -m pip install -r requirements-env.txt (from the repository root).
env_example = find_dotenv(".env.example", usecwd=True)
env_path = env_example.removesuffix(".example")
if not env_path or not os.path.isfile(env_path):
    raise FileNotFoundError("Create a local .env using the root .env.example first.")
load_dotenv(env_path)
```

```python

from pymongo.mongo_client import MongoClient
from pymongo.server_api import ServerApi

uri = os.environ["MONGODB_URI_2"]

# Create a new client and connect to the server
client = MongoClient(uri, server_api=ServerApi('1'))

# Send a ping to confirm a successful connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
db = client.get_database('sample_mflix')
```

```python
db.list_collection_names()
```

```python
movies = db['movies']
```
