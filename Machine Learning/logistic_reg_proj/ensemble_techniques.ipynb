```python
import pandas as pd  # loading and manupulating data
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
# Load the data
X = pd.read_csv(r"D:\Data Science_bat\Machine Learning\linear_regression_project\raw_data\indian_engineering_student_placement.csv")
y = pd.read_csv(r"D:\Data Science_bat\Machine Learning\linear_regression_project\raw_data\placement_targets.csv")
```

```python
X = X.drop(['Student_ID','extracurricular_involvement'], axis='columns')
y = y.drop(['Student_ID',  "salary_lpa"], axis=1)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test,y_train,y_test = train_test_split(X, y)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,f1_score,precision_score,recall_score

def evaluate_model(model_name, y_true, y_pred):
    """Return the most common binary-classification metrics."""
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, pos_label="Placed", zero_division=0),
        "Recall": recall_score(y_true, y_pred, pos_label="Placed", zero_division=0),
        "F1 Score": f1_score(y_true, y_pred, pos_label="Placed", zero_division=0),
        "confusion_matrix": confusion_matrix(y_true, y_pred)
    }
```

```python
Dtree_model = DecisionTreeClassifier(class_weight= "balanced")
Dtree_model.fit(X_train,y_train)
Dtree_pred = Dtree_model.predict(X_test)
```
