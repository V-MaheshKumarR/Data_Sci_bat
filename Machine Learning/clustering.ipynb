```python
import pandas as pd

df_mall = pd.read_csv("Mall_Customers.csv",index_col="CustomerID")
df_mall.head(3)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 

plt.figure(1 , figsize = (10 , 5) )
sns.scatterplot(
    data=df_mall,
    x="Spending Score (1-100)",
    y="Annual Income (k$)",
    hue="Gender",
    size="Age",
    palette="Set2"
);
```

```python
from sklearn import preprocessing

X = df_mall.drop(["Gender","Age"],axis=1)
X_norm = preprocessing.normalize(X)
```

```python
import numpy as np
from sklearn.cluster import KMeans

inertia_list = []

for i in range(1,10):
    model = KMeans(n_clusters=i, init='k-means++')
    model.fit(X_norm)

    inertia_list.append(model.inertia_)
```

```python
sns.lineplot(inertia_list)
```

```python
model = KMeans(n_clusters=4)
model.fit(X_norm)
```

```python
model.labels_
```

```python
# pd.DataFrame()
df_mall['clusters'] = model.labels_
```

```python
df_mall
```

```python
sns.scatterplot(data=df_mall, x= 'Annual Income (k$)', y='Spending Score (1-100)', hue='clusters')
```

```python
from sklearn.cluster import DBSCAN



db_scan_model = DBSCAN(eps=0.045, min_samples=4, leaf_size=10)
db_scan_model.fit(X_norm)
db_scan_model.labels_
```

```python
# pd.DataFrame()
df_mall['dbscan_clusters'] = db_scan_model.labels_
```

```python
sns.scatterplot(data=df_mall, x= 'Annual Income (k$)', y='Spending Score (1-100)', hue='dbscan_clusters')
```
