## 1. Common OTC antipyretics (fever‑reducers) in India

## 2. How to pick a tablet (if you’re otherwise healthy)

1. **First line:** **Paracetamol** (Crocin, Panadol, Dolo‑650, etc.)  
   - Works well for mild‑moderate fever and aches.  
   - Gentle on the stomach; preferred if you have a history of gastritis or are on blood thinners.

2. **If pain is more intense** (e.g., sore throat, muscle aches) **or if paracetamol alone isn’t enough**, you can add **ibuprofen** (Brufen, Nims) **or use a fixed‑dose combo** (Combiflam).  
   - Take ibuprofen with food to protect the stomach.  
   - Do **not** combine ibuprofen with other NSAIDs (e.g., aspirin, naproxen).

3. **Avoid**:  
   - Taking *both* paracetamol and ibuprofen at the **exact same time** unless you’re using a combination product that’s formulated for it. You can stagger them (e.g., paracetamol now, ibuprofen 3 hrs later) if you need prolonged relief, but keep total daily limits in mind.  
   - Exceeding the maximum daily dose of either drug—overdose can cause serious liver (paracetamol) or kidney/gastric (ibuprofen) injury.

---

## 3. Practical steps you can take right now

| Step | What to do |
|------|------------|
| **1. Measure your temperature** | Use a digital thermometer. Fever is generally considered ≥ 100.4 °F (38 °C). |
| **2. Hydrate** | Sip water, oral rehydration solution (ORS), clear soups, or coconut water. Fever increases fluid loss. |
| **3. Rest** | Give your body a chance to fight whatever is causing the fever. |
| **4. Choose a medication** | If you have no contraindications, start with **paracetamol 500 mg–650 mg** (1–2 tablets) now, then repeat every 4–6 hrs if needed, staying below 4 g per day. |
| **5. If paracetamol isn’t enough** | After 4–6 hrs, you can add **ibuprofen 400 mg** (1–2 tablets) **with food**, as long as you haven’t already taken another NSAID. |
| **6. Monitor** | If the temperature stays above 101 °F (38.5 °C) after 24 hrs of appropriate antipyretics, or if you develop new symptoms (e.g., rash, breathing difficulty, persistent vomiting, confusion), seek medical care. |
| **7. When to see a doctor** | • Fever > 104 °F (40 °C) <br>• Fever lasting > 3 days <br>• Accompanied by severe headache, stiff neck, severe abdominal pain, chest pain, or urinary symptoms <br>• Underlying chronic disease, pregnancy, or immunosuppression |

| **What if I have a stomach ulcer?** | Stick with **paracetamol only**. Avoid ibuprofen/NSAIDs unless a doctor says otherwise. |
| **I’m diabetic; any special concerns?** | Paracetamol is fine. Ibuprofen can affect kidney function and may raise blood pressure; use only if needed and under supervision. |
| **Can I give these to a child?** | Children need weight‑based dosing (e.g., paracetamol 10–15 mg/kg). Use pediatric formulations (syrup or chewable tablets) and follow the label or pediatrician’s guidance. |
