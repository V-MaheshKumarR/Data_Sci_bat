```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Simulate rolling two dice 10,000 times
# np.random.randint(1, 7) generates numbers 1-6
roll1 = np.random.randint(1, 7, 10000)
roll2 = np.random.randint(1, 7, 10000)
sums = roll1 + roll2

# Plot the distribution
plt.figure(figsize=(10, 6))
sns.histplot(sums, bins=np.arange(1.5, 13.5, 1), kde=True, color='skyblue')
plt.title('Dice Roll Distribution (2d6)')
plt.xlabel('Sum of Dice')
plt.ylabel('Frequency')
plt.show()   
```

## Statistics for Data Science Beginners

### Slide 1: Why Statistics Matters
- Data science starts with data
- Statistics helps us summarize and understand information
- It helps us make decisions using evidence

### Slide 2: What is Data?
- Numerical data: age, marks, salary
- Categorical data: gender, city, favorite subject
- Data is the raw material of data science

### Slide 3: Mean, Median, Mode
- Mean = average value
- Median = middle value when sorted
- Mode = most frequent value

Example: 10, 12, 12, 15, 18, 20, 25
- Mean = 16
- Median = 15
- Mode = 12

### Slide 4: Variance and Standard Deviation
- Variance tells us how spread out the data is
- Standard deviation is the square root of variance
- Low SD = values close to the mean
- High SD = values far from the mean

### Slide 5: Distributions
- Distribution shows how values are spread
- Common patterns: normal, skewed, uniform

### Slide 6: Normal Distribution
- Bell-shaped curve
- Symmetric around the mean
- Mean = Median = Mode
- Many real-world measurements follow this pattern

### Slide 7: Standard Normal Distribution
- Mean = 0
- Standard deviation = 1
- Also called Z-distribution
- Used for z-scores and hypothesis testing

### Slide 8: Probability Basics
- Probability is the chance of an event
- Range: 0 to 1
- Example: coin toss has 50% chance of heads

### Slide 9: Sampling Techniques
- Simple random sampling
- Stratified sampling
- Systematic sampling
- Cluster sampling
- Samples are used when studying the whole population is not practical

### Slide 10: Hypothesis Testing
- Null hypothesis (H0): no difference or no effect
- Alternative hypothesis (H1): there is a difference or effect
- Used to decide if results are meaningful

### Slide 11: Z-score
- Formula: z = (x - mean) / standard deviation
- Tells how far a value is from the mean in standard deviation units

### Slide 12: Z-test
- Used when population standard deviation is known
- Compares sample mean with expected mean
- Helps decide if the difference is statistically significant

### Slide 13: Real-World Applications
- Education: compare student performance
- Healthcare: analyze patient results
- Business: evaluate sales and campaigns
- Machine learning: understand feature distributions

### Slide 14: Summary
- Statistics is the foundation of data science
- It helps us understand data and draw conclusions
- It supports better decisions and stronger models
