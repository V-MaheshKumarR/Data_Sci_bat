# Deep Learning Model - Save & Load Demo
Train a simple neural network (Keras), save it, load it back, and make predictions.
Similar to the ML model saving workflow but using TensorFlow/Keras.

```python
# Import Required Libraries
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
print(f"TensorFlow version: {tf.__version__}")
```

```python
# Load MNIST handwritten digits dataset
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

# Normalize pixel values to [0, 1]
X_train = X_train / 255.0
X_test = X_test / 255.0

# One-hot encode labels
y_train_cat = keras.utils.to_categorical(y_train, 10)
y_test_cat = keras.utils.to_categorical(y_test, 10)

print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
```

```python
# Build a Sequential model with Dense layers
model = keras.Sequential([
    layers.Flatten(input_shape=(28, 28)),       # flatten 28x28 images to 784
    layers.Dense(128, activation='relu'),        # hidden layer 1
    layers.Dense(64, activation='relu'),         # hidden layer 2
    layers.Dense(10, activation='softmax')       # output layer (10 digits)
])

model.summary()
```

```python
# Compile the model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(
    X_train, y_train_cat,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)
```

```python
# Evaluate on test set
test_loss, test_acc = model.evaluate(X_test, y_test_cat)
print(f"\nTest Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

```python
# Plot training & validation loss and accuracy
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(history.history['loss'], label='Train Loss')
ax1.plot(history.history['val_loss'], label='Val Loss')
ax1.set_title('Loss over Epochs')
ax1.set_xlabel('Epoch')
ax1.legend()

ax2.plot(history.history['accuracy'], label='Train Accuracy')
ax2.plot(history.history['val_accuracy'], label='Val Accuracy')
ax2.set_title('Accuracy over Epochs')
ax2.set_xlabel('Epoch')
ax2.legend()

plt.tight_layout()
plt.show()
```

## 6. Save the Trained Model
Keras supports multiple save formats:
- **`.keras`** — recommended native Keras format
- **`.h5`** — legacy HDF5 format
- **SavedModel directory** — TensorFlow's default format

```python
# Save model in .keras format (recommended)
model.save('mnist_model.keras')
print("Model saved as 'mnist_model.keras'")

# Save model in .h5 format (legacy)
model.save('mnist_model.h5')
print("Model saved as 'mnist_model.h5'")
```

## 7. Load the Saved Model & Make Predictions

```python
# Load the saved model
loaded_model = keras.models.load_model('mnist_model.keras')
print("Model loaded successfully!")

# Make predictions on test data
predictions = loaded_model.predict(X_test[:5])
predicted_labels = np.argmax(predictions, axis=1)
actual_labels = y_test[:5]

print(f"\nPredicted: {predicted_labels}")
print(f"Actual:    {actual_labels}")
```

```python
# Display sample predictions with images
fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for i, ax in enumerate(axes):
    ax.imshow(X_test[i], cmap='gray')
    ax.set_title(f"Pred: {predicted_labels[i]} | True: {actual_labels[i]}")
    ax.axis('off')
plt.tight_layout()
plt.show()
```
