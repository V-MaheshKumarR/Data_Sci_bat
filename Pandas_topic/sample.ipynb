```python
charges_info =  insurance_df['charges'].describe()
charges_info
```

```python
Q1 = charges_info['25%']
Q3 = charges_info['75%']

IQR = Q3 - Q1
print(IQR)

lower_boundry = Q1 - (1.5 * IQR)
upper_boundry = Q3 + (1.5 * IQR)
```

```python
lower_boundry, upper_boundry
```

```python
outlier_removed_df = insurance_df[(insurance_df['charges'] < upper_boundry) & (insurance_df['charges'] > lower_boundry)]
```

```python
outlier_removed_df.duplicated().sum()
```

```python
outlier_removed_df.drop_duplicates(inplace=True)
```
