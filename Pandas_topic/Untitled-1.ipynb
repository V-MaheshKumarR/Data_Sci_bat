**What is EDA?**
Exploratory Data Analysis is the *first* thing we do after getting data. Before any
machine-learning model, we must answer:

1. **What is in the data?** — rows, columns, data types
2. **Is the data clean?** — missing values, wrong dtypes, duplicates
3. **How is each column distributed?** — central tendency, spread, outliers
4. **How do columns relate?** — categories vs numbers

## 2. Creating Sample Data with Random Numbers

Before touching real data, we practise on **synthetic data** because:

* we already know the "truth" (marks must lie between 45 and 100),
* there are no missing values or dirty entries to distract us,
* it is reproducible and instant — no file needed.

`random.randint(45, 100)` returns an integer **inclusive** of both ends.
`range(1, 200)` runs 199 times, so we get **199 marks** (not 200 — a classic
off-by-one point to highlight in class).

```python
random.seed(42)  # makes the "random" numbers repeatable for the whole class

marks = [random.randint(45, 100) for i in range(1, 200)]

print("Number of marks generated:", len(marks))
marks[:15]  # peek at the first 15
```

```python
# Convert the plain Python list into a pandas Series
ser = pd.Series(data=marks)

ser.head()
```

> **A Series is a single column with an index.**
> A DataFrame is simply many Series glued together sharing one index.

## 3. Descriptive Statistics on a Pandas Series

`describe()` returns **eight numbers** that summarise a whole numeric column:

| Statistic | Meaning |
|---|---|
| `count` | Number of **non-null** values |
| `mean`  | Arithmetic average — sensitive to outliers |
| `std`   | Standard deviation — average distance from the mean (spread) |
| `min`   | Smallest value |
| `25%`   | **Q1** — 25% of values are below this |
| `50%`   | **Median (Q2)** — the middle value |
| `75%`   | **Q3** — 75% of values are below this |
| `max`   | Largest value |

**Quick reading tips**
* `mean ≈ median` → roughly symmetric distribution.
* `mean > median` → right-skewed (a few large values pulling the mean up).
* `max` far away from `75%` → suspect outliers.

```python
ser.describe()
```

```python
# Individual statistics can also be pulled out one by one
print("Mean   :", ser.mean())
print("Median :", ser.median())
print("Mode   :", ser.mode().tolist())
print("Std    :", ser.std())
print("Range  :", ser.max() - ser.min())
```

## 4. Quick Plotting of a Series

`ser.plot()` uses the **line plot** as its default. A line plot joins point *i* to
point *i+1*, so it only makes sense when the order of the rows carries meaning
(time series, sequence, sorted values).

Our marks are in **random order**, so the line is just noise. For unordered
values, prefer:

* **histogram / KDE** → shape of the distribution
* **boxplot** → median, quartiles, outliers

```python
ser.plot(figsize=(10, 3), title="Default line plot (order is meaningless here)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(ser, bins=15, kde=True, ax=axes[0])
axes[0].set_title("Histogram — better for unordered marks")

sns.boxplot(x=ser, orient="h", ax=axes[1])
axes[1].set_title("Boxplot — spread and outliers")

plt.tight_layout()
plt.show()
```
